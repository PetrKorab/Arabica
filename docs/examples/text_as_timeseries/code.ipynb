```python
import pandas as pd
from arabica import arabica_freq
```

```python
data = pd.read_csv('headlines.csv')
```

```python
head = data.head(7)
styled_head = head.style.set_table_styles(
    [{'selector': 'th', 'props': [('text-align', 'left'), ('font-size', '12pt')]}, #colums
     {'selector': 'td', 'props': [('text-align', 'left'), ('font-size', '10pt')]}] # text
)
display(styled_head)
```

```python
result = arabica_freq(text = data['headline'], # Text
             time = data['date'],              # Time
             time_freq = 'M',                  # Aggregation period
             max_words = 4,                    # Max number for n-grams to be displayed
             date_format = 'us',               # Use US-style date format to parse dates
             stopwords = ['english'],          # Language for stop words
             stopwords_ext = None,             # Remove extended list of English stopwords
             skip = ['Jan','Feb','Dr',         # Remove additional strings
                     "—","Says"],
             lower_case = True,                # Lowercase text before cleaning and frequency analysis
             numbers = True)                   # Remove all digits
```

```python
styled_head = result.style.set_table_styles(
    [{'selector': 'th', 'props': [('text-align', 'left'), ('font-size', '12pt')]}, #colums
     {'selector': 'td', 'props': [('text-align', 'left'), ('font-size', '10pt')]}] # text
)
display(styled_head)
```

```python
arabica_freq(text = data['headline'],    # Text
             time = data['date'],        # Time
             time_freq = 'ungroup',      # Aggregation period
             max_words = 10,             # Max number for n-grams to be displayed
             date_format = 'us',         # Use US-style date format to parse dates
             stopwords = ["English"],    # Language for stop words
             stopwords_ext = None,       # Remove extended list of English stopwords
             skip = ['Jan','Feb','Dr',   # Remove additional strings
                      "—","Says"],
             lower_case = False,         # Lowercase text before cleaning and frequency analysis
             numbers = True)             # Remove all digits
```

```python
styled_head = result.style.set_table_styles(
    [{'selector': 'th', 'props': [('text-align', 'left'), ('font-size', '12pt')]}, #colums
     {'selector': 'td', 'props': [('text-align', 'left'), ('font-size', '10pt')]}] # text
)
display(styled_head)
```
